## 2. Mental State Classification System

```python
# Mental state classification overview
print("Mental State Classification System")
print("=" * 40)

mental_states = {
    "Relaxed (State 0)": {
        "description": "Calm, neutral emotional states",
        "eeg_characteristics": "High alpha (8-13 Hz), low beta (13-30 Hz)",
        "voice_emotions": ["calm", "neutral"],
        "typical_scenarios": ["Meditation", "Rest", "Light reading", "Peaceful environment"],
        "recommendations": ["Maintain current state", "Continue relaxing activities"]
    },
    "Focused (State 1)": {
        "description": "Alert, positive, engaged states",
        "eeg_characteristics": "High beta (13-30 Hz), moderate alpha, elevated gamma (30+ Hz)",
        "voice_emotions": ["happy", "surprise"],
        "typical_scenarios": ["Problem solving", "Learning", "Creative work", "Engaged conversation"],
        "recommendations": ["Optimal for challenging tasks", "Maintain focus with breaks"]
    },
    "Stressed (State 2)": {
        "description": "Anxious, fearful, negative states",
        "eeg_characteristics": "Very high beta, low alpha, elevated gamma, increased theta",
        "voice_emotions": ["angry", "fear", "sad"],
        "typical_scenarios": ["High pressure", "Anxiety", "Frustration", "Overwhelming tasks"],
        "recommendations": ["Stress reduction techniques", "Breathing exercises", "Take breaks"]
    }
}

for state, info in mental_states.items():
    print(f"\n{state}:")
    print(f"  Description: {info['description']}")
    print(f"  EEG Pattern: {info['eeg_characteristics']}")
    print(f"  Voice Emotions: {', '.join(info['voice_emotions'])}")
    print(f"  Scenarios: {', '.join(info['typical_scenarios'])}")

# Visualize state characteristics
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# EEG frequency band patterns
states = ['Relaxed', 'Focused', 'Stressed']
alpha_levels = [30, 18, 10]  # Typical alpha power levels
beta_levels = [10, 25, 40]   # Typical beta power levels
gamma_levels = [5, 12, 25]   # Typical gamma power levels

x = np.arange(len(states))
width = 0.25

axes[0].bar(x - width, alpha_levels, width, label='Alpha (8-13 Hz)', alpha=0.8)
axes[0].bar(x, beta_levels, width, label='Beta (13-30 Hz)', alpha=0.8)
axes[0].bar(x + width, gamma_levels, width, label='Gamma (30+ Hz)', alpha=0.8)

axes[0].set_title('EEG Frequency Patterns by Mental State', fontweight='bold')
axes[0].set_xlabel('Mental State')
axes[0].set_ylabel('Relative Power Level')
axes[0].set_xticks(x)
axes[0].set_xticklabels(states)
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# State transition flow
transition_matrix = np.array([
    [0.7, 0.2, 0.1],  # From Relaxed
    [0.3, 0.6, 0.1],  # From Focused  
    [0.2, 0.3, 0.5]   # From Stressed
])

im = axes[1].imshow(transition_matrix, cmap='Blues', alpha=0.8)
axes[1].set_title('State Transition Probabilities', fontweight='bold')
axes[1].set_xlabel('To State')
axes[1].set_ylabel('From State')
axes[1].set_xticks(range(3))
axes[1].set_yticks(range(3))
axes[1].set_xticklabels(states)
axes[1].set_yticklabels(states)

# Add text annotations
for i in range(3):
    for j in range(3):
        axes[1].text(j, i, f'{transition_matrix[i, j]:.1f}', 
                    ha='center', va='center', fontweight='bold')

plt.colorbar(im, ax=axes[1], shrink=0.6)
plt.tight_layout()
plt.show()

print("\n✓ Mental state classification overview complete")
```

## 3. Feature Engineering Pipeline

```python
# Feature engineering overview
print("Advanced Feature Engineering Pipeline")
print("=" * 45)

feature_categories = {
    "Time-Domain Features (120+)": [
        "Statistical moments (mean, std, skewness, kurtosis)",
        "Hjorth parameters (activity, mobility, complexity)",
        "Zero crossings and slope changes",
        "Energy and power measures",
        "Percentiles and range statistics"
    ],
    "Frequency-Domain Features (200+)": [
        "Power spectral density analysis",
        "Standard EEG bands (delta, theta, alpha, beta, gamma)",
        "Relative band powers and ratios",
        "Spectral entropy and centroid",
        "Peak frequencies and rolloff points"
    ],
    "Wavelet Features (300+)": [
        "Multi-level discrete wavelet transform",
        "Wavelet packet decomposition",
        "Energy distribution across scales",
        "Wavelet entropy measures",
        "Time-frequency localization"
    ],
    "Nonlinear Features (150+)": [
        "Sample and approximate entropy",
        "Permutation entropy",
        "Fractal dimensions (Higuchi, Katz, Petrosian)",
        "Lyapunov exponents",
        "Detrended fluctuation analysis"
    ],
    "Connectivity Features (100+)": [
        "Cross-channel correlations",
        "Coherence analysis",
        "Phase synchronization",
        "Graph theory measures",
        "Network topology features"
    ],
    "Harmonic Features (60+)": [
        "Harmonic-to-noise ratios",
        "Fundamental frequency tracking",
        "Harmonic structure analysis",
        "Pitch-related features",
        "Spectral harmonicity"
    ]
}

total_features = 0
for category, features in feature_categories.items():
    feature_count = int(category.split('(')[1].split('+')[0])
    total_features += feature_count
    print(f"\n{category}:")
    for feature in features:
        print(f"  • {feature}")

print(f"\nTotal Features: {total_features}+ comprehensive EEG features")

# Visualize feature distribution
categories = [cat.split(' (')[0] for cat in feature_categories.keys()]
counts = [int(cat.split('(')[1].split('+')[0]) for cat in feature_categories.keys()]

plt.figure(figsize=(12, 8))

# Pie chart
plt.subplot(2, 1, 1)
colors = plt.cm.Set3(np.linspace(0, 1, len(categories)))
wedges, texts, autotexts = plt.pie(counts, labels=categories, autopct='%1.1f%%', 
                                  colors=colors, startangle=90)
plt.title('Feature Distribution by Category', fontsize=14, fontweight='bold')

# Bar chart
plt.subplot(2, 1, 2)
bars = plt.bar(categories, counts, color=colors, alpha=0.8)
plt.title('Feature Count by Category', fontsize=14, fontweight='bold')
plt.xlabel('Feature Category')
plt.ylabel('Number of Features')
plt.xticks(rotation=45, ha='right')
plt.grid(True, alpha=0.3)

# Add value labels on bars
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, 
             str(count), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

print("\n✓ Feature engineering pipeline overview complete")
```
